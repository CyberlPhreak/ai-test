```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Data analysis notebook is working")
```

```python
data = {
    "month": ["January", "February", "March", "April", "May", "June"],
    "sales": [1200, 1500, 1800, 1700, 2100, 2500],
    "expenses": [800, 950, 1100, 1050, 1300, 1500]
}

df = pd.DataFrame(data)
df["profit"] = df["sales"] - df["expenses"]

df
```

```python
df.describe()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df["month"], df["sales"], marker="o", label="Sales")
plt.plot(df["month"], df["expenses"], marker="o", label="Expenses")
plt.plot(df["month"], df["profit"], marker="o", label="Profit")

plt.title("Sales, Expenses and Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.legend()
plt.grid(True)
plt.show()
```
